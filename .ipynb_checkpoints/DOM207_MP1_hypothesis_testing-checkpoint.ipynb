# DOM207 MP1 — Statistical Analysis & Hypothesis Testing

This notebook picks up after `data_preprocessing_nb.ipynb` (cleaning) and `DOM207_MP1_eda_visualizations.ipynb` (EDA/plots). It runs on **`Restaurant_Cleaned.csv`** and contains all formal hypothesis tests (HT) and correlation/association analysis for the report's *Analysis* section.

**Workflow for every test below:**
1. State H0 / H1
2. Check assumptions (normality via Shapiro-Wilk, equal variance via Levene's)
3. Pick the appropriate test based on what the assumption checks show (don't force a test)
4. Report the test statistic, p-value, and an effect size
5. Interpret in plain business language


```python
import pandas as pd
import numpy as np
from scipy import stats
import itertools

pd.set_option("display.precision", 4)
ALPHA = 0.05
```

## 1. Load cleaned data

We reload `Restaurant_Cleaned.csv` independently (rather than trusting notebook state) and recreate the two engineered ratio columns from the EDA notebook (`tip_pct`, `spend_per_person`) so this file can be run standalone.

```python
df = pd.read_csv("Restaurant_Cleaned.csv")

# keep day ordering consistent with the EDA notebook
day_order = ["thursday", "friday", "saturday", "sunday"]
df["day"] = pd.Categorical(df["day"].str.lower(), categories=day_order, ordered=True)

df["smoker"] = df["smoker"].astype(bool)
df["tip_pct"] = (df["tip"] / df["amount"]) * 100
df["spend_per_person"] = df["amount"] / df["partysize"]

print(df.shape)
df.head()
```

## 2. Assumption checking helper

A small helper that runs Shapiro-Wilk normality tests on each group and Levene's test for equal variances across groups, so we can justify picking a parametric (t-test/ANOVA) vs. non-parametric (Mann-Whitney/Kruskal-Wallis) test for every comparison, instead of assuming one.

```python
def check_assumptions(groups: dict, label: str):
    """groups: {group_name: pd.Series of values}. Prints Shapiro-Wilk per group + Levene's test across groups."""
    print(f"--- Assumption check: {label} ---")
    normal_flags = []
    for name, vals in groups.items():
        stat, p = stats.shapiro(vals)
        normal = p > ALPHA
        normal_flags.append(normal)
        print(f"  Shapiro-Wilk [{name}] (n={len(vals)}): W={stat:.4f}, p={p:.4g} -> {'normal' if normal else 'NOT normal'}")

    lev_stat, lev_p = stats.levene(*groups.values())
    equal_var = lev_p > ALPHA
    print(f"  Levene's test (equal variance): stat={lev_stat:.4f}, p={lev_p:.4g} -> "
          f"{'equal variance' if equal_var else 'UNEQUAL variance'}")

    all_normal = all(normal_flags)
    return all_normal, equal_var

def rank_biserial(u_stat, n1, n2):
    """Effect size for Mann-Whitney U: r = 1 - 2U / (n1*n2); ranges -1..1."""
    return 1 - (2 * u_stat) / (n1 * n2)

def cohend(a, b):
    n1, n2 = len(a), len(b)
    pooled_sd = np.sqrt(((n1 - 1) * a.std(ddof=1) ** 2 + (n2 - 1) * b.std(ddof=1) ** 2) / (n1 + n2 - 2))
    return (a.mean() - b.mean()) / pooled_sd

results = []  # collects a summary row per test for the final table
```

## HT1 — Bill amount: Lunch vs. Dinner

**H0:** μ(amount | lunch) = μ(amount | dinner)
**H1:** μ(amount | lunch) ≠ μ(amount | dinner)

Business question: does the average bill size differ by meal period? This matters for staffing and pricing decisions.

```python
lunch = df.loc[df.time == "lunch", "amount"]
dinner = df.loc[df.time == "dinner", "amount"]

all_normal, equal_var = check_assumptions({"lunch": lunch, "dinner": dinner}, "amount by time")

if all_normal:
    t_stat, p_val = stats.ttest_ind(lunch, dinner, equal_var=equal_var)
    test_name = "Welch t-test" if not equal_var else "Independent t-test"
    effect = cohend(lunch, dinner)
    effect_label = "Cohen's d"
else:
    t_stat, p_val = stats.mannwhitneyu(lunch, dinner, alternative="two-sided")
    test_name = "Mann-Whitney U"
    effect = rank_biserial(t_stat, len(lunch), len(dinner))
    effect_label = "rank-biserial r"

decision = "Reject H0" if p_val < ALPHA else "Fail to reject H0"
print(f"{test_name}: statistic={t_stat:.4f}, p={p_val:.4g}, {effect_label}={effect:.3f}")
print(f"Decision (alpha={ALPHA}): {decision}")
print(f"Lunch mean=${lunch.mean():.2f} (n={len(lunch)})  |  Dinner mean=${dinner.mean():.2f} (n={len(dinner)})")

results.append(dict(test="HT1: Amount ~ Time", method=test_name, statistic=t_stat, p_value=p_val,
                     effect_size=effect, effect_label=effect_label, decision=decision))
```

## HT2 — Bill amount: Smoker vs. Non-smoker parties

**H0:** μ(amount | smoker) = μ(amount | non-smoker)
**H1:** μ(amount | smoker) ≠ μ(amount | non-smoker)

```python
smoker_amt = df.loc[df.smoker == True, "amount"]
nonsmoker_amt = df.loc[df.smoker == False, "amount"]

all_normal, equal_var = check_assumptions({"smoker": smoker_amt, "non-smoker": nonsmoker_amt}, "amount by smoker")

if all_normal:
    t_stat, p_val = stats.ttest_ind(smoker_amt, nonsmoker_amt, equal_var=equal_var)
    test_name = "Welch t-test" if not equal_var else "Independent t-test"
    effect = cohend(smoker_amt, nonsmoker_amt)
    effect_label = "Cohen's d"
else:
    t_stat, p_val = stats.mannwhitneyu(smoker_amt, nonsmoker_amt, alternative="two-sided")
    test_name = "Mann-Whitney U"
    effect = rank_biserial(t_stat, len(smoker_amt), len(nonsmoker_amt))
    effect_label = "rank-biserial r"

decision = "Reject H0" if p_val < ALPHA else "Fail to reject H0"
print(f"{test_name}: statistic={t_stat:.4f}, p={p_val:.4g}, {effect_label}={effect:.3f}")
print(f"Decision (alpha={ALPHA}): {decision}")
print(f"Smoker mean=${smoker_amt.mean():.2f} (n={len(smoker_amt)})  |  Non-smoker mean=${nonsmoker_amt.mean():.2f} (n={len(nonsmoker_amt)})")

results.append(dict(test="HT2: Amount ~ Smoker", method=test_name, statistic=t_stat, p_value=p_val,
                     effect_size=effect, effect_label=effect_label, decision=decision))
```

## HT3 — Relationship between Amount and Tip

**H0:** ρ(amount, tip) = 0 (no association)
**H1:** ρ(amount, tip) ≠ 0

Since `amount` failed normality above (and both variables are on the same skewed scale), Spearman is the primary result; Pearson is reported alongside for reference/comparison since the assignment sample uses continuous ratio data.

```python
pearson_r, pearson_p = stats.pearsonr(df["amount"], df["tip"])
spearman_r, spearman_p = stats.spearmanr(df["amount"], df["tip"])

print(f"Pearson r = {pearson_r:.4f}, p = {pearson_p:.4g}")
print(f"Spearman rho = {spearman_r:.4f}, p = {spearman_p:.4g}")

decision = "Reject H0" if spearman_p < ALPHA else "Fail to reject H0"
print(f"Decision (alpha={ALPHA}, based on Spearman): {decision}")

results.append(dict(test="HT3: Amount vs Tip", method="Spearman", statistic=spearman_r, p_value=spearman_p,
                     effect_size=spearman_r, effect_label="rho (effect size)", decision=decision))
```

## HT4 — Relationship between Party Size and Amount

**H0:** ρ(partysize, amount) = 0
**H1:** ρ(partysize, amount) ≠ 0

Note: `partysize` was capped at its 99th percentile during cleaning, and Pearson vs. Spearman can diverge noticeably when the relationship is monotonic but non-linear — worth flagging in the report if you see that here.

```python
pearson_r2, pearson_p2 = stats.pearsonr(df["partysize"], df["amount"])
spearman_r2, spearman_p2 = stats.spearmanr(df["partysize"], df["amount"])

print(f"Pearson r = {pearson_r2:.4f}, p = {pearson_p2:.4g}")
print(f"Spearman rho = {spearman_r2:.4f}, p = {spearman_p2:.4g}")

decision = "Reject H0" if spearman_p2 < ALPHA else "Fail to reject H0"
print(f"Decision (alpha={ALPHA}, based on Spearman): {decision}")

if abs(spearman_r2 - pearson_r2) > 0.15:
    print("Note: Pearson and Spearman diverge notably here -> relationship is likely monotonic but non-linear, "
          "not just 'weak'. Worth a scatterplot with a lowess/trend line in the report.")

results.append(dict(test="HT4: Partysize vs Amount", method="Spearman", statistic=spearman_r2, p_value=spearman_p2,
                     effect_size=spearman_r2, effect_label="rho (effect size)", decision=decision))
```

## HT5 — Association between Day and Time (Chi-square test of independence)

**H0:** Day of week and meal time (lunch/dinner) are independent
**H1:** Day of week and meal time are NOT independent

Business relevance: if certain days skew heavily lunch or dinner, that directly informs staffing schedules and targeted promotions.

```python
contingency = pd.crosstab(df["day"], df["time"])
print(contingency)

chi2, p_val, dof, expected = stats.chi2_contingency(contingency)

# Cramer's V effect size
n = contingency.sum().sum()
min_dim = min(contingency.shape) - 1
cramers_v = np.sqrt((chi2 / n) / min_dim)

decision = "Reject H0" if p_val < ALPHA else "Fail to reject H0"
print(f"\nChi-square = {chi2:.4f}, dof = {dof}, p = {p_val:.4g}")
print(f"Cramer's V = {cramers_v:.3f}")
print(f"Decision (alpha={ALPHA}): {decision}")

results.append(dict(test="HT5: Day vs Time (chi-square)", method="Chi-square", statistic=chi2, p_value=p_val,
                     effect_size=cramers_v, effect_label="Cramer's V", decision=decision))
```

## HT6 (bonus) — Does average bill amount differ across Days?

**H0:** μ(amount | thursday) = μ(amount | friday) = μ(amount | saturday) = μ(amount | sunday)
**H1:** at least one day's mean differs

Four groups → ANOVA if normal/equal variance, Kruskal-Wallis otherwise, with pairwise post-hoc comparisons (Bonferroni-corrected Mann-Whitney) if the omnibus test is significant.

```python
day_groups = {name: sub["amount"].values for name, sub in df.groupby("day", observed=True)}

print("--- Assumption check: amount by day ---")
normal_flags = []
for name, vals in day_groups.items():
    stat, p = stats.shapiro(vals)
    normal_flags.append(p > ALPHA)
    print(f"  Shapiro-Wilk [{name}] (n={len(vals)}): p={p:.4g} -> {'normal' if p > ALPHA else 'NOT normal'}")
lev_stat, lev_p = stats.levene(*day_groups.values())
print(f"  Levene's test: stat={lev_stat:.4f}, p={lev_p:.4g}")

if all(normal_flags) and lev_p > ALPHA:
    f_stat, p_val = stats.f_oneway(*day_groups.values())
    omnibus_name = "One-way ANOVA"
    stat_val = f_stat
else:
    h_stat, p_val = stats.kruskal(*day_groups.values())
    omnibus_name = "Kruskal-Wallis"
    stat_val = h_stat

k = len(day_groups)
n_total = sum(len(v) for v in day_groups.values())
eta_sq = (stat_val - k + 1) / (n_total - k) if omnibus_name == "Kruskal-Wallis" else None

decision = "Reject H0" if p_val < ALPHA else "Fail to reject H0"
print(f"\n{omnibus_name}: statistic={stat_val:.4f}, p={p_val:.4g}")
if eta_sq is not None:
    print(f"eta-squared (effect size) = {eta_sq:.3f}")
print(f"Decision (alpha={ALPHA}): {decision}")

results.append(dict(test="HT6: Amount ~ Day (omnibus)", method=omnibus_name, statistic=stat_val, p_value=p_val,
                     effect_size=eta_sq, effect_label="eta-squared", decision=decision))

# Post-hoc pairwise comparisons (only meaningful if omnibus test rejected H0)
if p_val < ALPHA:
    print("\nPost-hoc pairwise Mann-Whitney (Bonferroni-corrected):")
    pairs = list(itertools.combinations(day_groups.keys(), 2))
    bonferroni_alpha = ALPHA / len(pairs)
    for d1, d2 in pairs:
        u, p = stats.mannwhitneyu(day_groups[d1], day_groups[d2], alternative="two-sided")
        sig = "significant" if p < bonferroni_alpha else "not significant"
        print(f"  {d1} vs {d2}: U={u:.1f}, p={p:.4g} -> {sig} (adj. alpha={bonferroni_alpha:.4g})")
```

## Bonus — Tip percentage by waiter gender

Not required, but directly useful for the recommendations section: does the served party tip a higher **percentage** depending on the waiter's gender? (Recall `gender` is the *waiter's* gender per the spec, not the customer's.)

**H0:** μ(tip_pct | male waiter) = μ(tip_pct | female waiter)
**H1:** μ(tip_pct | male waiter) ≠ μ(tip_pct | female waiter)

```python
male_tip = df.loc[df.gender == "male", "tip_pct"]
female_tip = df.loc[df.gender == "female", "tip_pct"]

all_normal, equal_var = check_assumptions({"male": male_tip, "female": female_tip}, "tip_pct by gender")

if all_normal:
    t_stat, p_val = stats.ttest_ind(male_tip, female_tip, equal_var=equal_var)
    test_name = "Welch t-test" if not equal_var else "Independent t-test"
    effect = cohend(male_tip, female_tip)
    effect_label = "Cohen's d"
else:
    t_stat, p_val = stats.mannwhitneyu(male_tip, female_tip, alternative="two-sided")
    test_name = "Mann-Whitney U"
    effect = rank_biserial(t_stat, len(male_tip), len(female_tip))
    effect_label = "rank-biserial r"

decision = "Reject H0" if p_val < ALPHA else "Fail to reject H0"
print(f"{test_name}: statistic={t_stat:.4f}, p={p_val:.4g}, {effect_label}={effect:.3f}")
print(f"Decision (alpha={ALPHA}): {decision}")
print(f"Male waiter mean tip%={male_tip.mean():.2f}% (n={len(male_tip)}) | "
      f"Female waiter mean tip%={female_tip.mean():.2f}% (n={len(female_tip)})")

results.append(dict(test="Bonus: Tip% ~ Waiter Gender", method=test_name, statistic=t_stat, p_value=p_val,
                     effect_size=effect, effect_label=effect_label, decision=decision))
```

## Summary table

A single table of every test run above — statistic, p-value, effect size, and decision — ready to drop straight into the report's *Analysis* section.

```python
summary = pd.DataFrame(results)
summary["p_value"] = summary["p_value"].map(lambda x: f"{x:.4g}")
summary["statistic"] = summary["statistic"].map(lambda x: f"{x:.4f}")
summary["effect_size"] = summary["effect_size"].map(lambda x: f"{x:.3f}" if pd.notna(x) else "-")
summary
```

## Notes for the Discussion/Recommendations section

- Every group here failed the Shapiro-Wilk normality test (unsurprising — `amount`/`tip` are right-skewed financial data), which is why Mann-Whitney/Kruskal-Wallis/Spearman are used as the **primary** decision basis rather than the t-test/ANOVA/Pearson equivalents. Report both if your instructor wants to see the parametric version too, but justify the non-parametric choice explicitly in the write-up — that justification is itself part of what's being graded.
- Effect sizes are included because a significant p-value with n=357 doesn't by itself tell you whether a difference is *practically* meaningful for a business recommendation — pair every "reject H0" with its effect size when you write the Discussion.
- Cramer's V and eta-squared translate roughly as: ~0.1 small, ~0.3 medium, ~0.5 large.
