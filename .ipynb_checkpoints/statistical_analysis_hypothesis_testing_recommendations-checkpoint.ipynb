# Statistical Analysis, Hypothesis Testing & Business Recommendations

**Dataset:** `Restaurant_Cleaned.csv`  
**Purpose:** Perform data-supported hypothesis tests and translate the statistically supported findings into business recommendations.

### Approach
The tests below are selected from the variables available in the cleaned dataset rather than applying tests arbitrarily.

- **HT 1 — Lunch vs Dinner:** Does the average bill amount differ between lunch and dinner?
- **HT 2 — Smoker vs Non-Smoker:** Does the average bill amount differ between smoker and non-smoker tables?
- **HT 3 — Party Size vs Bill Amount:** Is there a statistically significant monotonic relationship between party size and bill amount?
- **HT 4 — Bill Amount vs Tip:** Is there a statistically significant relationship between bill amount and tip?
- **HT 5 — Day × Time:** Are day of week and meal period associated?

Significance level for all tests: **α = 0.05**.


```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("Restaurant_Cleaned.csv")

# Derived variables used in the analysis
df["tip_pct"] = (df["tip"] / df["amount"]) * 100
df["spend_per_person"] = df["amount"] / df["partysize"]

print("Shape:", df.shape)
display(df.head())
```

## 1. Preliminary Statistical Checks

The cleaned dataset contains 357 observations. The normality results from the EDA notebook indicate that the main financial variables are not normally distributed. Therefore:

- For the two-group bill comparisons, **Welch's independent two-sample t-test** is used because it does not require equal variances and is reasonably robust with these group sizes.
- For relationships involving party size, **Spearman correlation** is used because party size is discrete and the relationship is better assessed as a monotonic association.
- Pearson correlation is also reported for bill amount vs tip because it is useful for describing the linear relationship.
- The categorical Day × Time question is tested using **Chi-square independence**.


```python
alpha = 0.05

summary = pd.DataFrame({
    "Metric": ["Total observations", "Mean bill", "Median bill", "Mean tip"],
    "Value": [
        len(df),
        df["amount"].mean(),
        df["amount"].median(),
        df["tip"].mean()
    ]
})

display(summary)
```

# HT 1 — Lunch vs Dinner

### Question
Is the mean bill amount different between lunch and dinner?

**H₀:** μ_lunch = μ_dinner  
**H₁:** μ_lunch ≠ μ_dinner

A two-sided Welch t-test is used.


```python
lunch = df.loc[df["time"] == "lunch", "amount"]
dinner = df.loc[df["time"] == "dinner", "amount"]

test1 = stats.ttest_ind(lunch, dinner, equal_var=False)

# 95% CI for (Lunch mean - Dinner mean)
def welch_ci(a, b, confidence=0.95):
    a, b = np.asarray(a), np.asarray(b)
    na, nb = len(a), len(b)
    ma, mb = a.mean(), b.mean()
    va, vb = a.var(ddof=1), b.var(ddof=1)

    diff = ma - mb
    se = np.sqrt(va / na + vb / nb)
    df_w = (va / na + vb / nb) ** 2 / (
        (va / na) ** 2 / (na - 1) + (vb / nb) ** 2 / (nb - 1)
    )
    critical = stats.t.ppf((1 + confidence) / 2, df_w)
    return diff, diff - critical * se, diff + critical * se, df_w

diff1, ci1_low, ci1_high, df1 = welch_ci(lunch, dinner)

ht1 = pd.DataFrame({
    "Group": ["Lunch", "Dinner"],
    "N": [len(lunch), len(dinner)],
    "Mean Bill": [lunch.mean(), dinner.mean()],
    "Median Bill": [lunch.median(), dinner.median()],
    "Std Dev": [lunch.std(), dinner.std()]
})

display(ht1)

print(f"Welch t-statistic = {test1.statistic:.4f}")
print(f"p-value = {test1.pvalue:.6f}")
print(f"Mean difference (Lunch - Dinner) = {diff1:.2f}")
print(f"95% CI for mean difference = ({ci1_low:.2f}, {ci1_high:.2f})")

if test1.pvalue < alpha:
    print("Decision: Reject H0.")
    print("Conclusion: The data provide statistically significant evidence that mean bill amount differs between lunch and dinner.")
else:
    print("Decision: Fail to reject H0.")
    print("Conclusion: The data do not provide statistically significant evidence that mean bill amount differs between lunch and dinner.")
```

### HT 1 interpretation

The observed dinner mean is higher than the lunch mean, but the p-value is above 0.05 and the confidence interval for the mean difference includes zero. Therefore, the observed difference should **not** be treated as statistically established in this sample.

**Business implication:** Do not recommend a dinner-only expansion or pricing change solely because the sample dinner mean is higher.


# HT 2 — Smoker vs Non-Smoker

### Question
Is the mean bill amount different between smoker and non-smoker tables?

**H₀:** μ_smoker = μ_non-smoker  
**H₁:** μ_smoker ≠ μ_non-smoker

A two-sided Welch t-test is used.


```python
smoker = df.loc[df["smoker"] == True, "amount"]
non_smoker = df.loc[df["smoker"] == False, "amount"]

test2 = stats.ttest_ind(smoker, non_smoker, equal_var=False)
diff2, ci2_low, ci2_high, df2 = welch_ci(smoker, non_smoker)

ht2 = pd.DataFrame({
    "Group": ["Smoker", "Non-Smoker"],
    "N": [len(smoker), len(non_smoker)],
    "Mean Bill": [smoker.mean(), non_smoker.mean()],
    "Median Bill": [smoker.median(), non_smoker.median()],
    "Std Dev": [smoker.std(), non_smoker.std()]
})

display(ht2)

print(f"Welch t-statistic = {test2.statistic:.4f}")
print(f"p-value = {test2.pvalue:.6f}")
print(f"Mean difference (Smoker - Non-Smoker) = {diff2:.2f}")
print(f"95% CI for mean difference = ({ci2_low:.2f}, {ci2_high:.2f})")

if test2.pvalue < alpha:
    print("Decision: Reject H0.")
    print("Conclusion: The data provide statistically significant evidence that mean bill amount differs by smoker status.")
else:
    print("Decision: Fail to reject H0.")
    print("Conclusion: The data do not provide statistically significant evidence that mean bill amount differs by smoker status.")
```

### HT 2 interpretation

Smoker tables have a higher observed mean bill in this sample, but the difference is not statistically significant at α = 0.05.

**Business implication:** Smoker status should not be used as a standalone basis for targeting higher-spending customers from this dataset.


# HT 3 — Party Size vs Bill Amount

### Question
Is there a statistically significant monotonic relationship between party size and total bill amount?

**H₀:** ρ = 0  
**H₁:** ρ ≠ 0

Because party size is discrete and the relationship need not be strictly linear, **Spearman's rank correlation** is used.


```python
rho, p3 = stats.spearmanr(df["partysize"], df["amount"])

print(f"Spearman rho = {rho:.4f}")
print(f"p-value = {p3:.10f}")

if p3 < alpha:
    print("Decision: Reject H0.")
    print("Conclusion: There is a statistically significant positive monotonic relationship between party size and bill amount.")
else:
    print("Decision: Fail to reject H0.")
    print("Conclusion: There is insufficient evidence of a monotonic relationship.")
```

### HT 3 interpretation

The Spearman correlation is positive and statistically significant. Larger parties tend to be associated with larger total bills.

This does **not** mean that increasing party size automatically causes a higher bill; it describes an association in the observed data.


# HT 4 — Bill Amount vs Tip

### Question
Is there a statistically significant relationship between total bill amount and tip?

**H₀:** ρ = 0  
**H₁:** ρ ≠ 0

Pearson correlation is used to quantify the linear association, while Spearman correlation is also reported as a robustness check.


```python
pearson_r, pearson_p = stats.pearsonr(df["amount"], df["tip"])
spearman_r, spearman_p = stats.spearmanr(df["amount"], df["tip"])

ht4 = pd.DataFrame({
    "Method": ["Pearson", "Spearman"],
    "Correlation": [pearson_r, spearman_r],
    "p-value": [pearson_p, spearman_p]
})

display(ht4)

if pearson_p < alpha:
    print("Decision for Pearson test: Reject H0.")
    print("Conclusion: Bill amount and tip have a statistically significant positive linear relationship.")
else:
    print("Decision for Pearson test: Fail to reject H0.")
```

### HT 4 interpretation

Bill amount and tip show a statistically significant positive relationship. The positive association is also supported by the Spearman result.

**Business implication:** As customer bills increase, absolute tip amounts generally increase as well. This is useful for understanding tip-volume patterns, although it should not be interpreted as proof of causation.


# HT 5 — Day × Time Chi-square Test

### Question
Are the day of week and meal period independent?

**H₀:** Day and time are independent.  
**H₁:** Day and time are associated.

A Chi-square test of independence is appropriate because both variables are categorical.


```python
contingency = pd.crosstab(df["day"], df["time"])

chi2, p5, dof, expected = stats.chi2_contingency(contingency)

print("Observed counts:")
display(contingency)

expected_df = pd.DataFrame(
    expected,
    index=contingency.index,
    columns=contingency.columns
)

print("Expected counts under independence:")
display(expected_df.round(2))

print(f"Chi-square statistic = {chi2:.4f}")
print(f"Degrees of freedom = {dof}")
print(f"p-value = {p5:.10f}")

if p5 < alpha:
    print("Decision: Reject H0.")
    print("Conclusion: Day of week and meal period are statistically associated in this dataset.")
else:
    print("Decision: Fail to reject H0.")
    print("Conclusion: There is insufficient evidence of an association between day and meal period.")
```

### HT 5 interpretation

The Chi-square test indicates a statistically significant association between day and meal period. The observed distribution is therefore not consistent with day and meal period being independent in this sample.

This is primarily an **operational pattern**, not evidence that one meal period is inherently more profitable.


# Overall Statistical Findings

| Question | Test | p-value | Result |
|---|---|---:|---|
| Lunch vs Dinner bill | Welch two-sample t-test | 0.1144 | Not statistically significant |
| Smoker vs Non-Smoker bill | Welch two-sample t-test | 0.1615 | Not statistically significant |
| Party size vs bill | Spearman correlation | < 0.0001 | Significant positive association |
| Bill amount vs tip | Pearson correlation | < 0.0001 | Significant positive association |
| Day × Time | Chi-square independence | < 0.0001 | Significant association |

**Important:** “Fail to reject H₀” does not prove that the groups are identical. It means the sample does not provide sufficient evidence of a difference at α = 0.05.


# Business Recommendations

The recommendations below are directly tied to the statistically supported findings.

### 1. Use party size when planning capacity and revenue expectations
**Finding:** Party size has a statistically significant positive monotonic relationship with total bill amount.

**Proposition:** Use expected party-size mix as one input when planning table capacity, seating allocation and revenue expectations. Larger groups can generate larger total bills, so operational planning should account for the expected mix of group sizes.

### 2. Use bill size as an indicator of absolute tip volume
**Finding:** Bill amount and tip have a statistically significant positive relationship.

**Proposition:** When forecasting tip volume or staffing-related service outcomes, bill volume can be used as an informative predictor of absolute tips. This should be treated as an association rather than a causal relationship.

### 3. Plan operations around the observed day–meal-period pattern
**Finding:** Day and meal period are statistically associated.

**Proposition:** Staffing, table allocation and inventory planning can be reviewed by day and meal period rather than treating every shift as having the same customer mix.

### 4. Avoid unsupported segmentation decisions
**Finding:** The lunch-vs-dinner and smoker-vs-non-smoker mean bill differences are not statistically significant at α = 0.05.

**Proposition:** The current sample does not provide sufficient statistical evidence to justify making major targeting, pricing or expansion decisions solely around these two comparisons.

### Overall business proposition

**Based on the observed customer spending patterns, the restaurant should prioritize operational planning around party-size mix, day/meal-period demand patterns and bill-volume relationships with tipping, while avoiding strong segmentation decisions where the hypothesis tests do not show statistically significant differences.**


# Limitations

- The analysis is based on the cleaned dataset supplied for the assignment.
- Statistical significance does not establish causation.
- The data represent observed transactions; they should not automatically be generalized to all restaurant customers or future periods.
- The hypothesis tests use α = 0.05.
- The Day × Time association is descriptive of this dataset and should be validated on a larger or more representative time period before making major operational changes.
